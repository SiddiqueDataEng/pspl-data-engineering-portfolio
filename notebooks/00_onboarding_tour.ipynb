## What you are standing inside

Imagine a programme office asking: *Are cash transfers reaching people on time, which districts lag, and do donor commitments match spend?* This repo answers with **data engineering mechanics**, using **synthetic** Pakistani social protection and Afghan refugee-style datasets.

**The arc:** files → **Bronze Delta** (ingest) → **Silver Delta** (PySpark notebook) → **Gold marts in DuckDB** (dbt) → **KPI SQL + Streamlit**.

Open the documentation hub anytime: `docs/README.md`.

## Folder roles (mental map)

| Location | Role |
|----------|------|
| `data_large/` | Nine source files (CSV.gz, Parquet, JSON, Avro). |
| `ingest/` | Reads sources, writes **Bronze** Delta under `delta_lake/bronze/`. |
| `notebooks/delta_lake_operations.ipynb` | Cleans Bronze → **Silver** Delta under `delta_lake/silver/`. |
| `dbt/` | Builds `stg_*` → `int_*` → `mart_*`; persists to `pspl.duckdb`. |
| `sql/` | Seven standalone KPI queries against DuckDB. |
| `dashboard/streamlit_app.py` | Interactive charts over the same marts. |
| `docs/` | Learning guides, runbooks, HTML command helper. |

**Why this split:** Spark handles **lake-scale** Delta IO; DuckDB + dbt handles **fast, testable** modeling on a laptop.

## Run order (do not skip)

1. **venv + Java** — see `docs/LEARNING_GUIDE.md` Sections 4–7.
2. **Bronze:** `python ingest/ingest.py` or `scripts/run-full-pipeline.ps1`.
3. **Silver:** run `delta_lake_operations.ipynb` (nbconvert runs it in pipelines).
4. **Gold:** `cd dbt`, set `DELTA_LAKE_PATH` to your `delta_lake` folder with forward slashes, then `dbt run` and `dbt test`.
5. **Consume:** `sql/*.sql`, Streamlit (`scripts/run-dashboard.ps1`), or Jupyter.

If `dbt` complains about `DELTA_LAKE_PATH`, the Windows full pipeline script sets it automatically.
